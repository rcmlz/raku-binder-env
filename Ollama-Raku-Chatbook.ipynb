# Jupyter + Raku + Ollama

## Ollama
Install [Ollama](https://ollama.com/download) and pull one [model](https://ollama.com/search).

````raku
%%bash
ollama list
````

Make sure the model you choose is installed. Run e.g.

```bash
%%bash
ollama pull llama3.2
```

in a Jupyter code cell to download your favorite model.

````raku
my $model_name = 'llama3.2'
````

## Raku
We use the [Raku](https://raku.org/) package [LLM::Functions](https://raku.land/zef:antononcube/LLM::Functions) for interacting via Ollama with the Large Language Model (LLM) of your choice.

````raku
use LLM::Functions
````

Do you want to create an `ollama` configuration using default settings

````raku
my $llm-evaluator = llm-configuration('ollama')
````

or do prefer to adjust some of the settings?

````raku
my $llm-evaluator = llm-configuration('ollama', model => $model_name, base-url => 'http://127.0.0.1:11434')
````

Fancy some hand-crafted priming

````raku
my $priming = "You are the best Raku coder in the world and know Raku inside out. 
               You give short and informative answers.
               You code like a student in 10th grade."
````

or are you fine with a one of the pre-defined prompts from [LLM::Prompts](https://raku.land/zef:antononcube/LLM::Prompts)?

````raku
use LLM::Prompts
````

````raku
my $priming = llm-prompt('CodeWriterX')('Raku')
````

## Ask

Need some help in coding Raku?

````raku
my $prompt = 'Using the Raku programming language: How to read from STDIN and 
              apply a filter to every line removing all lines that start with "#" ?'
````

````raku
#% markdown
llm-synthesize([
      $priming,
      $prompt
    ],
    :$llm-evaluator
)
````

## Chat

Do you want a continuous chat?

````raku
sink my $priming = "You are the best Raku coder in the world and know Raku and
                    its documentation on https://docs.raku.org/ inside out. 
                    You give short and informative answers refering to build-in Raku functions using simple example code."
````

````raku
my $chat = llm-chat(chat-id => 'raku-expert-talk', conf => $llm-evaluator, prompt => $priming)
````

````raku
#% markdown
$chat.eval('What are the advantages of Raku over other common programming languages?')
````

Attention: (Simple) models hallucinate a lot!

````raku
#% markdown
$chat.eval('Explain the build-in Grammar class of Raku!')
````

That should not work with a simple no-frills local LLM - right?

````raku
#% markdown
$chat.eval('Explain the Raku Grammar example shown on https://raku.org/')
````
