## LLM-Setup

### Ollama Cloud Models

Ollama API keys can be set here: [ollama.com/settings/keys](https://ollama.com/settings/keys).

```raku
my $OLLAMA_API_KEY='put-your-key-here'
```

```raku
# Ollama Cloud Models
my $client = WWW::Ollama::Client.new(
     host => "https://ollama.com",
     api-key => $OLLAMA_API_KEY
)
```

Browse the available cloud models here: [docs.ollama.com/cloud](https://docs.ollama.com/cloud#cloud-models)

### Local Models
Alternatively [install Ollama](https://ollama.com/download) on your computer and use this local instance.

```raku
# Ollama on localhost on port 11434
my $client = WWW::Ollama::Client.new
```

or any instance in your local network.

```raku
# Ollama on other host in local network
my $client = WWW::Ollama::Client.new(
     host => "http://10.0.2.2",
     port => 11434
)
```

### Helperfunctions

We use `ask()` und `chat()` for simplified access.

```raku
#| Shows available models
sub show-models-from(WWW::Ollama::Client $client) {
   my @models = |$client.list-models;
   @models
    ==> { .sort(*<modified_at>.DateTime ) }()
    ==> { .map({ $_<modified_at> = $_<modified_at>.DateTime.Str.split('T').head; $_ }) }()
    ==> { .map({ $_<size-short> = round($_<size> / 1024 ** 3, 0.1) ~ ' GB'; $_ }) }()
    ==> { .map({ $_.grep(*.key ∈ <name size size-short modified_at>) })».Hash }()
}

#| Simple chat - no history
sub ask($prompt, :model_name(:$model)) {
    my $response = $client.completion($prompt, :$model);
    $response<content>
}

#| Simple conversation - incl. history
sub chat($prompt, :model_name(:$model)) {
    @messages.push({role => "user", content => $prompt});
    my $response = $client.chat({:@messages, :$model});
    @messages.push({role => "assistant", content => $response<content>});
    $response<content>
}
```

### Model selection
