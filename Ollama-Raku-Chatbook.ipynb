# Jupyter + Raku + Ollama

Install [Ollama](https://ollama.com/download) and one of the available [models](https://ollama.com/search).

```raku
my $model_name = 'llama3'
```

We use the [Raku](https://raku.org/) package [LLM::Functions](https://raku.land/zef:antononcube/LLM::Functions) for interacting via Ollama with the Large Language Model of your choice.

```raku
use LLM::Functions
```

Create a configuration using `ollama` as template for unspecified parameters.

```raku
my $my-ollama-conf = llm-configuration('ollama', model => $model_name, base-url => 'http://127.0.0.1:11434');
```

Prefer some priming?

```raku
my $priming = "You are the best Raku coder in the world and know Raku inside out. 
               You give short and informative answers.
               You code like a student in 10th grade."
```

Need some help?

```raku
my $prompt = 'Using the Raku programming language: How to read from STDIN and apply a filter to every line red?'
```

```raku
#% markdown
llm-synthesize([
      $priming,
      $prompt
    ],
    e => $my-ollama-conf
);
```
